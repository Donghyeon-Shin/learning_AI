```python
import os
import numpy as np
from PIL import Image

## 1) isbi dataset 경로
isbi_dir = '../../Data/isbi-2012'

## 2) 데이터 불러오기
name_label = 'train-labels.tif'
name_image = 'train-volume.tif'

img_label = Image.open(os.path.join(isbi_dir, name_label))
img_input = Image.open(os.path.join(isbi_dir, name_image))

## 3) 데이터 확인
print(img_label.size)
print(img_input.size)

num_frame = img_label.n_frames # 총 프레임 수

## 4) training, valdation, test 데이터 분할 경로 설정
num_frame_train = 24
num_frame_val = 3
num_frame_test = 3

dir_train_image = os.path.join(isbi_dir, 'train/images')
dir_train_label = os.path.join(isbi_dir, 'train/labels')

dir_val_image = os.path.join(isbi_dir, 'val/images')
dir_val_label = os.path.join(isbi_dir, 'val/labels')

dir_test_image = os.path.join(isbi_dir, 'test/images')
dir_test_label = os.path.join(isbi_dir, 'test/labels')


## 디렉토리 생성
os.makedirs(dir_train_image, exist_ok=True)
os.makedirs(dir_train_label, exist_ok=True)
os.makedirs(dir_val_image, exist_ok=True)
os.makedirs(dir_val_label, exist_ok=True)
os.makedirs(dir_test_image, exist_ok=True)
os.makedirs(dir_test_label, exist_ok=True)


## 5) 데이터 shuffle
id_frame = np.arange(num_frame)
np.random.shuffle(id_frame)
```

```python
offset_frame = 0

# training data 추출
for i in range(num_frame_train):
    frame_id = id_frame[i + offset_frame]
    img_input.seek(frame_id)
    img_label.seek(frame_id)

    label = np.asarray(img_label)
    input = np.asarray(img_input)

    np.save(os.path.join(dir_train_label, f'{i:03d}.npy'), label)
    np.save(os.path.join(dir_train_image, f'{i:03d}.npy'), input)

offset_frame += num_frame_train

# validation data 추출
for i in range(num_frame_val):
    frame_id = id_frame[i + offset_frame]
    img_input.seek(frame_id)
    img_label.seek(frame_id)

    label = np.asarray(img_label)
    input = np.asarray(img_input)

    np.save(os.path.join(dir_val_label, f'{i:03d}.npy'), label)
    np.save(os.path.join(dir_val_image, f'{i:03d}.npy'), input)

offset_frame += num_frame_val

# test data 추출
for i in range(num_frame_test):
    frame_id = id_frame[i + offset_frame]
    img_input.seek(frame_id)
    img_label.seek(frame_id)

    label = np.asarray(img_label)
    input = np.asarray(img_input)

    np.save(os.path.join(dir_test_label, f'{i:03d}.npy'), label)
    np.save(os.path.join(dir_test_image, f'{i:03d}.npy'), input)
```
